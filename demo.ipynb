# Hopfield Network Memory Model
A demonstration of associative memory using a Hopfield Network. Store binary patterns and retrieve them from noisy inputs.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.hopfield.network import HopfieldNetwork
from src.hopfield.patterns import random_pattern, add_noise, pattern_overlap
```

## 1. Train the Network

```python
rng = np.random.default_rng(42)
net = HopfieldNetwork(100)
patterns = np.array([random_pattern(100, rng) for _ in range(5)])
net.train(patterns)
print(f'Trained on {len(patterns)} patterns with {net.n} neurons')
```

## 2. Recall from Noisy Input

```python
probe = add_noise(patterns[0], noise_ratio=0.20, rng=rng)
recalled = net.recall(probe, max_steps=500, rng=rng)
overlap = pattern_overlap(recalled, patterns[0])
print(f'Overlap with original: {overlap:.3f}')
```

## 3. Visualize Results

```python
fig, axes = plt.subplots(1, 3, figsize=(10, 4))
for ax, data, title in zip(axes, [patterns[0], probe, recalled], ['Original', 'Noisy (20%)', 'Recalled']):
    ax.imshow(data.reshape(10, 10), cmap='pink', vmin=-1, vmax=1)
    ax.set_title(title)
    ax.axis('off')
plt.suptitle(f'Pattern Recall — Overlap: {overlap:.3f}')
plt.tight_layout()
plt.show()
```
